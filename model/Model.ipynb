```python
from ultralytics import YOLO
```

```python
drive_save_dir='/content/drive/MyDrive/Car_detection_problem'
```

```python
model=YOLO('yolov8n.pt')
```

```python
results=model.train(data='/content/drive/MyDrive/Car_detection_problem/data.yaml',epochs=50,
                    imgsz=640,batch=16,project=drive_save_dir,name='Car Detection problem',save_period=5)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/drive/MyDrive/Car_detection_problem/Car Detection problem-2/results.csv')
```

```python
df.columns=df.columns.str.strip()
```

```python
plt.figure(figsize=(12,5))
# Box Loss
plt.subplot(1,2,1)
plt.plot(df['epoch'],df['train/box_loss'])
plt.plot(df['epoch'],df['val/box_loss'])
plt.xlabel('Epoch')
plt.ylabel('Box Loss')
plt.title('Box Loss')
plt.legend(['Train','Validation'])

# Class Loss
plt.subplot(1,2,2)
plt.plot(df['epoch'],df['train/cls_loss'])
plt.plot(df['epoch'],df['val/cls_loss'])
plt.xlabel('Epoch')
plt.ylabel('cls Loss')
plt.title('classification Loss')
plt.legend(['Train','Validation'])

plt.tight_layout()
plt.show()
```
